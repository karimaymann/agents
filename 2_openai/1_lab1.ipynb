```python
# The imports
import os
from dotenv import load_dotenv
from agents import Agent, Runner, trace, AsyncOpenAI, OpenAIChatCompletionsModel, ModelProvider

```

```python
# The usual starting point

load_dotenv(override=True)
```

```python
# Make an agent with name, instructions, model
gemini_client = AsyncOpenAI(base_url=os.getenv("GEMINI_BASE_URL"), api_key=os.getenv("GOOGLE_API_KEY"))
gemini_model = OpenAIChatCompletionsModel(model="gemini-2.0-flash", openai_client=gemini_client)
agent = Agent(name="Joker", instructions="You are the joker from The Dark Knight", model=gemini_model)
```

```python
# Run the joke with Runner.run(agent, prompt) then print final_outputS

with trace("Telling a joke"):
    result = await Runner.run(agent, "Talk about your scars.")
    print(result.final_output)
```

```python
gemini_client = AsyncOpenAI(base_url=os.getenv("GEMINI_BASE_URL"), api_key=os.getenv("GOOGLE_API_KEY"))
gemini_model = OpenAIChatCompletionsModel(model="gemini-2.0-flash", openai_client=gemini_client)

instructions = "You are the joker from The Dark Knight."

joker_agent = Agent(name="Joker", instructions=instructions, model=gemini_model)

result = await Runner.run(joker_agent, "What is your motive?")
result.final_output
```
